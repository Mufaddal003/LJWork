```python
#Given a list l of size m.you need to count the number of special elements in the given list.an element is special if removal of that element makes the list balanced.
# the list will be balanced if sum of even index is equal to sum of odd index elements.also display the updated lis after removal of special elements.
l = [5, 5, 2, 5, 8]
count = 0
new = []

for i in range(len(l)):
    s = l[:]
    a = s.pop(i)
    so = sum(s[x] for x in range(len(s)) if x % 2 != 0)
    se = sum(s[x] for x in range(len(s)) if x % 2 == 0)
    
    if so == se:
        count += 1
        print("Special Element =", a)
        new.append(a)

print("Count =", count)
print("Final List =", s)
            
```

```python
def calculate_pizza_price(size, toppings, cheese):
    cost = 0
    # Calculate pizza base price based on size
    if size == 'small':
        cost += 50
    elif size == 'medium':
        cost += 100
    else:
        cost += 200
    
    # Calculate topping prices
    topping_prices_20 = ['corn', 'tomato', 'onion', 'capsicum']
    topping_prices_50 = ['mushroom', 'olives', 'broccoli']
 
    for topping in toppings:
         cost += 20 if topping in topping_prices_20 else 50
    #Calculate cheese price
    cost += 50 * len(cheese)
    return cost

def order_pizza():
    pizzas = []
    number = int(input("How many pizzas you want to order: "))
    for i in range(number):
        toppings = []
        cheese = []
        print('Customize Pizza', i + 1)
        print('Select Pizza Size: small, medium or large')
        size = input('Select size: ')
        
        print('''Select Toppings: 
                20 for 'corn', 'tomato', 'onion', 'capsicum'
                50 for 'mushroom', 'olives', 'broccoli' ''')
        toppings_count = int(input('How many toppings: '))
    
        for _ in range(toppings_count):
             toppings.append(input('Enter toppings: '))
        
        print('Select Cheese: mozarrella, feta, cheddar')
        cheese_count = int(input('How many cheese: '))

        for _ in range(cheese_count):
             cheese.append(input('Enter cheese: '))
        print()
        pizza_cost = calculate_pizza_price(size, toppings, cheese)
        pizzas.append((size, toppings, cheese, pizza_cost))
    return pizzas

def print_bill(pizzas):
     total = 0
     count = 1
     for pizza in pizzas:
         print('Pizza', count)
         print("Size:", pizza[0])
         print("Toppings:", pizza[1])
         print("Cheese:", pizza[2])
         print("Pizza cost:", pizza[3])
         total += pizza[3]
         count += 1
     print('Total bill amount:', total)
# Main program
pizza_order = order_pizza()
print_bill(pizza_order)
```

```python
inventory={
"Laptop": {"price":55000,"quantity":8,"category":"Electronics"},
"Mobile": {"price":25000,"quantity":15,"category":"Electronics"},
"Chair": {"price":3500,"quantity":20,"category":"Furniture"},
"Table": {"price":7000,"quantity":10,"category":"Furniture"}
}
#1,2
total_cost={item: d["price"]*d["quantity"] for item,d in inventory.items()}
print(total_cost)
#3
high_value={item: v for item,v in total_cost.items() if v>200000}
print("Item EXCEDING 2Lakhs",high_value)
#4
categories={d["category"] for d in inventory.values()}
print("Unique Categories",categories)
#5
max_item=max(total_cost.items(),key=lambda kv:kv[1])
print(max_item)
#6
sorted_item= sorted(total_cost.items(),key=lambda kv:kv[1],reverse=True)
print("sorted items:")
for name,val in sorted_item:
    print(f"{name} Rs{val:}")
#7
category_total={c:sum(v for p,v in total_cost.items()
                     if inventory[p]["category"]==c)for c in categories}
print("Category Total",category_total)
```
