### Working With Files


1) Text File(.txt)
2) Binary File(we don't have to use this)
* File Access Modes
  1) Read Only(r)
  2) Read And write(r+)(aama aagal no data preserve thaase)
  3) Write only(w)
  4) Write and Read(w+)(aama aagal no data badho erase thai jaase and newly lakhvu parse)
  5) Append Only(a)
  6) Append and Read(a+)

```python
f=open('new.txt','r')#default mode is read
print(f.read())
print(f.closed)
f.close()
print(f.closed)
```

* Second Method To open File
* Main difference that u don't have to manually close the file in second method it will automatically close the file if we are out of intentation

```python
with open('new.txt','r') as f:
    print(f.read())
print(f.closed)
```

```python
#here first time the cursor will read full file and now in second time cursor position is at last so there is nothing to read 
f=open('new.txt','r')
print(f.tell())#shows the cursor position
print(f.read())
print(f.tell())
f.seek(10)#points the cursor to the appointated index
print(f.tell())
print(f.read())
f.close()
```

## Read And Write Mode
* Reading ANd Writing
* Handle position at begining of the file
* Raise an eroor if file doesn't exist

```python
with open('utf8_demo.txt','r+',encoding='utf-8') as f:
    print(f.read())
    f.write("\nThis is Demo file")
    f.seek(0)
    print(f.read())
```

```python
with open('utf8_demo.txt','r+',encoding='utf-8') as f:
    f.write("This is Demo file")#This will oveeride the existing content acording to the lenght
    f.seek(0)
    print(f.read())
```

## Write Only(w)
- Open Files for writing Only
- for the exisisting file,data is truncated and overwritten.
- the handle position at the begining of the file.
- creates the file if the file does not exist.

```python
with open('demo.txt','w',encoding='utf-8') as f:
    f.write("Hello\n")
    f.write("Hello 1")
    f.write("Hello 2")
```

## Write and Read(w+)
- Open Files for writing Only
- for the exisisting file,data is truncated and overwritten.
- the handle position at the begining of the file.
- creates the file if the file does not exist.

```python
with open('demo.txt','w+',encoding='utf-8') as f:
    f.write("Hello\n")
    f.write("Hello 1")
    f.write("Hello 2")
    f.seek(0)
    print(f.read())
```

## Appen Only(a)
- open the file for writing
- the file is created if it does not exist
- the handle is positioned at the end of the file
- the data being written will be inserted at the end after the existing data

```python
with open('deemo.txt','a',encoding='utf-8') as f:
    f.write("This Is Test")
```

## Append and Read(a+)


```python
with open('dmo.txt','a+',encoding='utf-8') as f:
    f.write("This Is Test")
    f.seek(0)
    print(f.read())
```

# Reading The Files
- **read()** - Returns read bytes in the form of a string.
- **readline()**- reads a line of the file and returns in the form of string.
- **readlines()**- reads all the lines of the file and return them as a string element in a list.

```python
with open('text.txt') as f:
    print(f.read())
    f.seek(0)
    print(f.read(2))
    print(f.read(6))
```
