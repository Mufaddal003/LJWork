## Write Only(w)
- Open Files for writing Only
- for the exisisting file,data is truncated and overwritten.
- the handle position at the begining of the file.
- creates the file if the file does not exist.

```python
with open('demo.txt','w',encoding='utf-8') as f:
    f.write("Hello\n")
    f.write("Hello 1")
    f.write("Hello 2")
```

## Write and Read(w+)
- Open Files for writing Only
- for the exisisting file,data is truncated and overwritten.
- the handle position at the begining of the file.
- creates the file if the file does not exist.

```python
with open('demo.txt','w+',encoding='utf-8') as f:
    f.write("Hello\n")
    f.write("Hello 1")
    f.write("Hello 2")
    f.seek(0)
    print(f.read())
```

## Appen Only(a)
- open the file for writing
- the file is created if it does not exist
- the handle is positioned at the end of the file
- the data being written will be inserted at the end after the existing data

```python
with open('deemo.txt','a',encoding='utf-8') as f:
    f.write("This Is Test")
```

```python
with open("newp.txt", "r+") as f:
    lines = f.readlines()
    f.seek(0)
    i = 0
    while i < len(lines):
        l = lines[i]
        i += 1
        if len(l) != 0:
            if l[0] == "#":
                continue
            else:
                if "#" in l:
                    for index in range(1, len(l)):
                        if l[index] == "#":
                            f.write(l[0:index] + "\n")
                            break
                else:
                    f.write(l)
        else:
            break
    f.truncate()
    f.seek(0)
    print(f.read())

            
```

```python
#WAP to read a text file and print the content of text file with word starting with I in reverse order
with open("text.txt",'r') as f:
    new=""
    r=f.read().split()
    for i in r:
        if i[0]=="I" or i[0]=='i':
            new+=i[::-1]+" "
        else:
            new+=i+" "
    print(new)
```

```python
#WAP to count the frequency in a word in given text file
with open("freq.txt", "r") as f:
    words = f.read().split()

freq = {}

for word in words:
    freq[word] = freq.get(word, 0) + 1

for word, count in freq.items():
    print(f"{word}: {count}")
            
```

```python
#SIR VERSION
with open("freq.txt", "r") as f:
    words = f.read().split()
    d={}
    for i in words:
        if i in d:
            d[i]+=1
        else:
            d[i]=1
    for i in d:
        print(i,"-",d[i])
```
